# 1. Imports
Make sure to install pandas, matplotlib, and seaborn before starting this notebook!

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# 2. Import CSV as DataFrame
The filepath for the CSV is "../data/StudentPerformanceFactors.csv".

(Why '../'? It tells the interpreter to go to src's parent folder before searching for the data folder. Without it, it'll try to look for data in src, which will cause a FileNotFoundError.)

```python
data = pd.read_csv('../data/StudentPerformanceFactors.csv')
print(data.head())
```

# Generate graphs from data
For now, I just put attendance-to-examscore as a placeholder.

```python
# Creating linear regression line for Attendance vs Exam Score
m, b = np.polyfit(data['Attendance'], data['Exam_Score'], 1)

# Labeling and plotting
label_text = f'$y = {m:.2f}x + {b:.2f}$'
ax = data.plot(x='Attendance', y='Exam_Score', kind='scatter')
ax.plot(data['Attendance'], m * data['Attendance'] + b, color='red', label=label_text)
plt.xlabel('Attendance (%)')
plt.ylabel('Exam Score')
plt.title('Attendance vs Exam Score')
plt.legend(loc='best')

# Save the plot
plt.savefig('../results/attendance_vs_exam_score.png')

# Show the plot
plt.show()
```
